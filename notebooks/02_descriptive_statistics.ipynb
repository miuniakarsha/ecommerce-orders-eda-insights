# Phase 3: Descriptive Statistics

## Purpose

This notebook calculates descriptive statistics for the verified retail orders dataset.

Project 1 focused on checking whether the data was clean and reliable.  
Project 2 focuses on Exploratory Data Analysis.

In this phase, descriptive statistics are used to understand the basic behavior of the dataset, such as average order value, median order value, minimum and maximum values, and standard deviation.

Main question:

What do the basic numbers tell us about customer orders, product prices, quantities, cart size, and revenue?

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/processed/ecommerce_orders_project2_verified.csv")
df.head() #loading verified dataset
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
numeric_columns = ["Quantity", "UnitPrice", "ItemsInCart", "TotalPrice"]

df[numeric_columns].head()
```

```python
descriptive_stats = df[numeric_columns].agg(
    ["count", "mean", "median", "min", "max", "std"]
).T

descriptive_stats = descriptive_stats.round(2)

descriptive_stats
```

```python
#add Q1, IQR, Q3
descriptive_stats["Q1"] = df[numeric_columns].quantile(0.25).round(2)
descriptive_stats["Q3"] = df[numeric_columns].quantile(0.75).round(2)
descriptive_stats["IQR"] = (descriptive_stats["Q3"] - descriptive_stats["Q1"]).round(2)

descriptive_stats
```

```python
#save
descriptive_stats.to_csv("../reports/descriptive_statistics_summary.csv")

print("Descriptive statistics summary saved successfully.")
```

```python
#main business FPIs calculating
total_orders = df["OrderID"].nunique()
total_customers = df["CustomerID"].nunique()
total_revenue = df["TotalPrice"].sum()
average_order_value = df["TotalPrice"].mean()
median_order_value = df["TotalPrice"].median()

business_kpis = pd.DataFrame({
    "Metric": [
        "Total Orders",
        "Total Customers",
        "Total Revenue",
        "Average Order Value",
        "Median Order Value"
    ],
    "Value": [
        total_orders,
        total_customers,
        round(total_revenue, 2),
        round(average_order_value, 2),
        round(median_order_value, 2)
    ]
})

business_kpis
```

```python
business_kpis.to_csv("../reports/business_kpis_summary.csv", index=False)

print("Business KPI summary saved successfully.")
```

```python
mean_median_comparison = pd.DataFrame({
    "Column": numeric_columns,
    "Mean": [round(df[col].mean(), 2) for col in numeric_columns],
    "Median": [round(df[col].median(), 2) for col in numeric_columns],
    "Difference": [round(df[col].mean() - df[col].median(), 2) for col in numeric_columns]
})

mean_median_comparison
```

```python
category_columns = ["Product", "PaymentMethod", "OrderStatus", "ReferralSource", "HasCoupon"]

for col in category_columns: #count category values
    print(f"\n{col} Count Summary")
    display(
        df[col]
        .value_counts()
        .reset_index()
        .rename(columns={col: "Category", "count": "Count"})
    )
```

```python
for col in category_columns:
    summary_table = (
        df[col]
        .value_counts()
        .reset_index()
        .rename(columns={col: "Category", "count": "Count"})
    )
    
    summary_table.to_csv(f"../reports/{col.lower()}_count_summary.csv", index=False)

print("Category count summaries saved successfully.")
```
